```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

data = pd.read_csv('/content/spambase.csv')

print(data.head())

if data.isnull().values.any():
  data = data.fillna(0)

df_x = data.iloc[:, :-1]
df_y = data.iloc[:, -1]

x_train, x_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.4, random_state=4)

sc = StandardScaler()
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)

perceptron = Perceptron()
perceptron.fit(x_train_scaled, y_train)

pred_train = perceptron.predict(x_train_scaled)
pred_test = perceptron.predict(x_test_scaled)

accuracy_train = accuracy_score(y_train, pred_train)
accuracy_test = accuracy_score(y_test, pred_test)

print("Accuracy of the training data: ", accuracy_train)
print("Accuracy of the testing data: ", accuracy_test)
```
