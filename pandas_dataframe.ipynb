### Generate City Temperature Data for a Week

```python
# Generate a date range for a week
dates = pd.date_range(start='2023-07-01', periods=7, freq='D')

# Generate random temperatures for a city (e.g., in Celsius)
# Let's make some temperatures high enough to potentially trigger a heat wave
np.random.seed(42) # for reproducibility
temperatures = pd.Series(np.random.uniform(25, 35, size=7), index=dates, name='Temperature_Celsius')

print("Weekly City Temperatures:")
print(temperatures)
```

### Identify Heat Waves

To identify a heat wave, we need to define criteria. Let's say a heat wave occurs when the temperature is above `30°C` for at least `3 consecutive days`.

```python
heat_wave_threshold = 30 # degrees Celsius
consecutive_days_for_heat_wave = 3

# Identify days above the threshold
above_threshold = temperatures > heat_wave_threshold

# Find consecutive days above the threshold
# We convert the boolean series to integers (True=1, False=0)
# Then we use `rolling` sum to find sequences of '1's
consecutive_above_threshold = above_threshold.astype(int).rolling(window=consecutive_days_for_heat_wave).sum()

# A heat wave occurs when the rolling sum equals the number of consecutive days required
heat_wave_starts = consecutive_above_threshold[consecutive_above_threshold == consecutive_days_for_heat_wave].index

print(f"Heat wave threshold: {heat_wave_threshold}°C for {consecutive_days_for_heat_wave} consecutive days.")
print("\nDays with temperature above threshold:")
print(above_threshold[above_threshold == True])

print("\nPotential Heat Wave Start Dates:")
if not heat_wave_starts.empty:
    for start_date in heat_wave_starts:
        # Find the actual end date of the heat wave segment
        end_date = start_date + pd.Timedelta(days=consecutive_days_for_heat_wave - 1)
        # Ensure the heat wave condition holds for the full duration
        if all(temperatures.loc[start_date:end_date] > heat_wave_threshold):
            print(f"Heat wave detected from {start_date.strftime('%Y-%m-%d')} to {end_date.strftime('%Y-%m-%d')}")
else:
    print("No heat waves detected based on the defined criteria.")
```

```python
heat_waves = temperatures[temperatures > 30]
print(heat_waves)
print(f"Minimum temperature of the week is {heat_waves.min()}°C")
print(f"Maximum temperature of the week is {heat_waves.max()}°C")
```
