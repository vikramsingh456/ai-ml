````python
import pandas as pd
import numpy as np

# Sample data for cities and delivery staff counts
city_resources = {
    'City': ['New York', 'London', 'Delhi', 'Tokyo'],
    'Delivery_Boys': [15, 10, 25, 12]
}
df_staff = pd.DataFrame(city_resources)

# Sample order data
order_data = {
    'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'City': ['New York', 'New York', 'London', 'London', 'Delhi', 'Delhi', 'Tokyo', 'Tokyo'],
    'Expected_Date': pd.to_datetime(['2024-05-01', '2024-05-01', '2024-05-02', '2024-05-02', '2024-05-03', '2024-05-03', '2024-05-04', '2024-05-04']),
    'Actual_Date': pd.to_datetime(['2024-05-01', '2024-05-03', '2024-05-02', '2024-05-02', '2024-05-05', '2024-05-06', '2024-05-04', '2024-05-04'])
}
df_orders = pd.DataFrame(order_data)

# Calculate delay in days
df_orders['Delay_Days'] = (df_orders['Actual_Date'] - df_orders['Expected_Date']).dt.days
df_orders['Is_Late'] = df_orders['Delay_Days'] > 0

# Aggregate data per city
city_performance = df_orders.groupby('City').agg(
    Total_Orders=('Order_ID', 'count'),
    Late_Orders=('Is_Late', 'sum')
).reset_index()

# Merge with staff data to analyze requirement
df_analysis = pd.merge(city_performance, df_staff, on='City')
df_analysis['Late_Ratio'] = df_analysis['Late_Orders'] / df_analysis['Total_Orders']

def check_requirement(row):
    if row['Late_Ratio'] > 0.3:
        return 'Hire More Delivery Boys'
    elif row['Late_Ratio'] == 0 and row['Delivery_Boys'] > 15:
        return 'Overstaffed'
    else:
        return 'Optimal'

df_analysis['Recommendation'] = df_analysis.apply(check_requirement, axis=1)

display(df_analysis)
````

### Vectorization vs. `apply()`

*   **Vectorized operations:** (like `df['A'] + df['B']`) happen in compiled C code. They are extremely fast.
*   **`apply()` or `map()`:** These are 'iterative'. Pandas passes each element back to the Python interpreter one by one. This makes them much slower for large datasets.

````python
import time

# Create a large sample DataFrame
large_df = pd.DataFrame({'Numbers': np.random.randn(100000)})

# 1. Vectorized approach
start = time.time()
result_vec = large_df['Numbers'] * 2
end_vec = time.time() - start

# 2. Using apply (Not vectorized)
start = time.time()
result_apply = large_df['Numbers'].apply(lambda x: x * 2)
end_apply = time.time() - start

print(f"Vectorized time: {end_vec:.5f} seconds")
print(f"Apply (loop) time: {end_apply:.5f} seconds")
print(f"Vectorization is roughly {int(end_apply/end_vec)}x faster!")
````

### Optimization Techniques

Instead of using `.apply(check_requirement)`, we can use `np.where` or `df.loc`. These are significantly faster because they work on the entire array at once.

````python
# 1. Using np.where (Highly Optimized Vectorization)
# Syntax: np.where(condition, value_if_true, value_if_false)

df_analysis['Recommendation_Fast'] = np.where(
    df_analysis['Late_Ratio'] > 0.3,
    'Hire More Delivery Boys',
    np.where((df_analysis['Late_Ratio'] == 0) & (df_analysis['Delivery_Boys'] > 15), 'Overstaffed', 'Optimal')
)

display(df_analysis[['City', 'Recommendation', 'Recommendation_Fast']])
````

### Fast Alternatives to `.apply()`

1.  **`np.where`**: Best for simple if-else logic. It's truly vectorized and works at C-speed.
2.  **`np.vectorize`**: A 'convenience' wrapper. While it doesn't provide the full speed of C, it is often faster and cleaner than a standard Python loop or `.apply()` for custom functions.

````python
import numpy as np

# --- Method 1: np.where (Fastest for conditions) ---
# We can nested np.where to handle multiple conditions
df_analysis['Status_Fast'] = np.where(
    df_analysis['Late_Ratio'] > 0.5,
    'Urgent Attention',
    np.where(df_analysis['Late_Ratio'] > 0, 'Monitor', 'Perfect')
)

# --- Method 2: np.vectorize (For custom functions) ---
def complex_logic(ratio, boys):
    # Imagine complex math here
    return "High Load" if (ratio * boys) > 5 else "Normal Load"

# Create a vectorized version of the function
v_func = np.vectorize(complex_logic)

# Apply it directly to columns
df_analysis['Load_Status'] = v_func(df_analysis['Late_Ratio'], df_analysis['Delivery_Boys'])

display(df_analysis[['City', 'Status_Fast', 'Load_Status']])
````
